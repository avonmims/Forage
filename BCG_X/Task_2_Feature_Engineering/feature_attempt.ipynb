```python
import pandas as pd
```

# DateTime Conversion

```python
# clean_data datetime parsing -> client_df
client_df = pd.read_csv('clean_data.csv', parse_dates=[5, 6, 7, 8])
display(client_df.head())
display(client_df.info())

# price_data datetime parsing -> price_df
price_df = pd.read_csv('price_data.csv', parse_dates=[1])
display(price_df.head())
display(price_df.info())
```

# Feature Engineering

1. off-peak price diff (dec - jan)
2. mid-peak price diff (dec - jan)
3. on-peak price diff (dec - jan)
4. peak spread dec (on peak - off peak)
5. peak spread jan (on peak -  off peak)
6. aggregatte diff (on + mid + off)_dec - (on + mid+ off)_jan

```python
# extract month from price_date
price_df['month'] = price_df['price_date'].dt.month

# numerical features for pivot table
features = ['price_off_peak_var', 'price_off_peak_fix', 'price_mid_peak_var', 'price_mid_peak_fix',
            'price_peak_var', 'price_peak_fix']

# conversion from long to wide format
monthly_data = price_df.pivot_table(
    index='id',
    columns='month',
    values=features,
)
```

- We have converted the **['price_date']** column to month only, seeing as *our data spans only 1 year*, this brings our price data to closely matching the *length* of our client df. With each row representing one client, and all price info being represented by numbered columns.

```python
#display(monthly_data.sample(15))
print(monthly_data.info())
print(f'\nNUM ROWS WITH AT LEAST ONE MISSING VALUE: {monthly_data.isna().any(axis=1).sum()}\n')

nan_ids = monthly_data[monthly_data.isna().any(axis=1)].index
missing = client_df[client_df['id'].isin(nan_ids)][['id', 'churn']]

print(f'\nMISSING VALUES COMPOSITION:\n{missing['churn'].value_counts()}\n')

monthly_data = monthly_data.dropna()

print(monthly_data.info())
print(f'\nNUM ROWS WITH AT LEAST ONE MISSING VALUE: {monthly_data.isna().any(axis=1).sum()}')
```

- The months of August and December have the highest number of non-null entries across all pricing variables. Pointing towards these months having the highest reliability and retention rates. With January and March having the lowest number of non-null entries, suggesting the inverse.

- Out of 106 rows with at least one missing value, 68 are active customers, and 20 are inactive. This leaves 18 observations that do not have an ID match with client data. This checks out, because client_data has 1,490 less ID's than price_data. Meaning some of the missing value observations are present only within price_data.

- Since the missing values that do match are majority active users, we will be removing all rows that have at least one missing entry. This will provide us with complete and reliable customer data for the entire observational window of 1 year.
